```python
n = int(input())
d = 2
flag = False

while d < n:
    if (n % d == 0):
        flag = True
    d = d + 1

if flag:
    print("n is not prime")
else:
    print("n is prime")
```

```python
# my try

n = int(input())
s = ""
if(n == 0):
    print("0")
else:
    while(n % 10 == 0):
        n = n // 10
    while(n != 0):
        k = n % 10
        s = s + str(k)
        n = n // 10
    print(s)
```

```python
# optimal solution without using array concept

num = int(input())
reversed_num = 0

while num != 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num //= 10

print(str(reversed_num))
```

Sum of evens and odds

```python
n = int(input())
k = n
sum_even = 0
sum_odd = 0

while(k != 0):
    digit = k % 10
    if(digit % 2 == 0):
        sum_even = sum_even + digit
    else:
        sum_odd = sum_odd + digit
    k = k // 10

print(sum_even,"",sum_odd)
```

```python
# prime or not

def isPrime(x):
    for i in range(2,x,1):
        if(x%i==0):
            break
    else:
        return True
    return False
```

Pair sum

```python
def pairSum(arr,n,x):
    count=0
    for i in range(n):
        for j in range(i+1,n):
            if(arr[i]+arr[j]==x):
                count=count+1
    return count

print("enter the array: ")
arr = list(map(int,input().split()))
print(arr)
print("\nenter the pair sum: ")
x = int(input())
print(x)
print("\nthe number of pairs = ",pairSum(arr,len(arr),x))
```

Triplet Sum

```python
def findTriplet(arr, n, x) :
    count=0
    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                if(arr[i]+arr[j]+arr[k]==x):
                    count+=1
    return count

print("enter the array: ")
arr = list(map(int,input().split()))
print(arr)
print("\nenter the triplet sum: ")
x = int(input())
print(x)
print("\nthe number of triplets = ",findTriplet(arr,len(arr),x))
```

```python
n = list(map(int, input().split(" ")))
def binarySearch(arr, length, x):
    arr.sort()
    s = 0
    e = length-1
    m = (s+e)//2
    
    while(s <= e):
        if(arr[m] == x):
            return True
        elif(x > arr[m]):
            s = m+1
            m = (s+e)//2
        elif(x < arr[m]):
            e = m-1
            m = (s+e)//2
    else:
        return False


print(f"The given array is:\n{n}")
n.sort()
print(f"The array in ascending order is:\n{n}")
print("Enter the number you want to search:\n")
x = int(input())
print(f"We need to search the element {x}.")

if(binarySearch(n, len(n), x)):
    print(f"The number {x} is present in the given array.")
else:
    print(f"The number {x} is not present in the array.")
```

```python
# recommended

def insertionSort(arr):
	n = len(arr) 
	
	if n <= 1:
		return arr

	for i in range(1, n):
		key = arr[i]
		j = i-1
		while j >= 0 and key < arr[j]:
			arr[j+1] = arr[j] 
			j -= 1
		arr[j+1] = key 
	return arr

n = list(map(int, input().split(" ")))
print(insertionSort(n))    
```

Sort 0 1 2 in one scan

```python
def sort012(arr,n):
    pz=0
    pt=n-1
    i=0
    while(i<=pt):
        if(arr[i]==0):
            arr[i], arr[pz] = arr[pz], arr[i]
            i+=1
            pz+=1
        elif(arr[i]==1):
            i+=1
        elif(arr[i]==2):
            arr[i], arr[pt] = arr[pt], arr[i]
            pt-=1
    return arr

n = list(map(int, input().split(" ")))
print(n)
print(sort012(n,len(n))) 
```

Replace character in string

```python
def replaceChar(s, c1, c2):
    temp = ""
    for i in s:
        if(i==c1):
            temp = temp + c2
        else:
            temp = temp + c1
    return temp

print(replaceChar("abacad",'a','e'))
```

count #vowels, #consonants, #digits, #special character in a string

```python
def countInString(str):
    v,c,d,s = 0,0,0,0
    for i in str:
        if((ord(i)>=65 and ord(i)<=90) or (ord(i)>=97 and ord(i)<=122)):
            if(ord(i) in [97,101,105,111,117,65,69,73,79,85]):
                v+=1
            else:
                c+=1
        elif(ord(i)>=48 and ord(i)<=57):
            d+=1
        else:
            s+=1
    return v,c,d,s
    
str = "abeioccd1234ABGHJ@!#$%123"
v,c,d,s = countInString(str)
print(v,c,d,s)
```

```python
# correct but not optimal

def isPermutation(string1, string2):
    flag=True
    
    if(len(string1)==len(string2)):
        for i in string1:
            if (i in string2) and (string1.count(i)==string2.count(i)):
                continue
            else:
                flag=False
        return flag
    else:
        return False

isPermutation("abb","babc")
```

Remove consecutive duplicates

```python
def removeConsecutiveDuplicates(string):
    n = len(string)
    k=0
    res=""
    for i in range(1,n):
        if(string[i]!=string[k]):
            res=res+string[k]
        k+=1
    res=res+string[n-1]
    return res

print(removeConsecutiveDuplicates("aaaabbbcccddddbbb"))
```

Reverse Each Word

```python
def reverseEachWord(string):
    a = string.split(" ")
    res=""
    for i in a:
        res = res + i[::-1] + " "
    
    return res

print(reverseEachWord("Hello I am Janya"))
```

```python
# okay solution but not optimal

def highestOccuringChar1(string):
    uniqList =[]
    for element in string:
        if element not in uniqList:
            uniqList.append(element)

    hFreqChar = uniqList[0]
    for i in uniqList:
        if(string.count(i)>string.count(hFreqChar)):
            hFreqChar = i
    return hFreqChar

print(highestOccuringChar1("acccbbaba"))
```

Row wise traversal

```python
a = [[1,2,3],[4,5,6]]
def printRowWise(a):
    ans = []
    rows = len(a)
    cols = len(a[0])
    
    for row in range(rows):
        for col in range(cols):
            ans.append(a[row][col])

    return ans

print(printRowWise(a))
```

Column wise traversal

```python
a = [[1,2,3],[4,5,6]]
def printColWise(a):
    ans = []
    rows = len(a)
    cols = len(a[0])
    
    for col in range(cols):
        for row in range(rows):
            ans.append(a[row][col])

    return ans

print(printRowWise(a))
```

Set matrix to zero

```python
# my try

a = [
    [0,2,3,10,11,0],
    [4,0,6,13,14,15],
    [7,8,9,17,18,19]
    ]

def setMatrixToZero(a):
    rows = len(a)
    cols = len(a[0])

    i,j = [],[]

    for row in range(rows):
        for col in range(cols):
            if(a[row][col] == 0):
                i.append(row)
                j.append(col)
    
    for row in range(rows):
        for col in range(cols):
            if(row in i or col in j):
                a[row][col] = 0


    return a

print(setMatrixToZero(a))
```

```python
a = [
    [0,2,3,10,11,0],
    [4,0,6,13,14,15],
    [7,8,9,17,18,19]
    ]

def rowWiseSum(mat, nRows, mCols):
    for row in range(nRows):
        sum = 0
        for col in range(mCols):
            sum += mat[row][col]
    
        print(sum, end=" ")

rowWiseSum(a,3,6)
```
